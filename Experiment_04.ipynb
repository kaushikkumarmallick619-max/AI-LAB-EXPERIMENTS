```python
 # Program
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 3), ('E', 6)],
    'C': [('F', 5)],
    'D': [('G', 2)],
    'E': [('G', 1)],
    'F': [('G', 2)],
    'G': []
}

# Heuristic values
heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 4,
    'E': 2,
    'F': 2,
    'G': 0
}


def astar(start, goal):
    open_list = [(start, 0)]
    came_from = {}
    g_cost = {start: 0}

    while open_list:
        # Select node with minimum f = g + h
        open_list.sort(
            key=lambda x: x[1] + heuristic[x[0]]
        )

        current = open_list.pop(0)[0]

        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]

            path.append(start)
            path.reverse()
            return path, g_cost[goal]

        for neighbor, cost in graph[current]:
            new_cost = g_cost[current] + cost

            if (neighbor not in g_cost or
                    new_cost < g_cost[neighbor]):
                g_cost[neighbor] = new_cost
                came_from[neighbor] = current

                open_list.append((neighbor, new_cost))

    return None, float('inf')


# Main Program
start = input("Enter the start node: ").upper()
goal = input("Enter the goal node: ").upper()

path, cost = astar(start, goal)

if path:
    print("Shortest path:", " -> ".join(path))
    print("Actual cost:", cost)
else:
    print("Path not found")
```
