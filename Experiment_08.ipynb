```python
from collections import deque

# Function to solve the Water Jug Problem using BFS
def water_jug_problem(jug1_capacity, jug2_capacity, target):
    visited = set()
    queue = deque()

    # Initial state: both jugs are empty
    queue.append((0, 0, []))

    while queue:
        jug1, jug2, path = queue.popleft()

        # Check if current state has been visited
        if (jug1, jug2) in visited:
            continue

        visited.add((jug1, jug2))

        # Add current state to the path
        path = path + [(jug1, jug2)]

        # Check if target is reached
        if jug1 == target or jug2 == target:
            return path

        # Generate all possible next states
        next_states = [
            # Fill jug1
            (jug1_capacity, jug2),

            # Fill jug2
            (jug1, jug2_capacity),

            # Empty jug1
            (0, jug2),

            # Empty jug2
            (jug1, 0),

            # Pour jug1 -> jug2
            (
                jug1 - min(jug1, jug2_capacity - jug2),
                jug2 + min(jug1, jug2_capacity - jug2)
            ),

            # Pour jug2 -> jug1
            (
                jug1 + min(jug2, jug1_capacity - jug1),
                jug2 - min(jug2, jug1_capacity - jug1)
            )
        ]

        # Add unvisited states to the deque
        for state in next_states:
            if state not in visited:
                queue.append((state[0], state[1], path))

    return None


# Main program
jug1 = int(input("Enter capacity of jug1: "))
jug2 = int(input("Enter capacity of jug2: "))
target = int(input("Enter target volume: "))

solution = water_jug_problem(jug1, jug2, target)

if solution:
    print("\nSteps to reach the target:")

    for step in solution:
        print(step)
else:
    print("No solution exists.")
```
